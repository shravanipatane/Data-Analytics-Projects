## 5. Experiment — Cleaning Customer Names

I noticed that customer names were not consistently formatted.

I experimented with string methods and found that `.str.title()`
was useful for capitalizing the first letter of each word.

```python
df["Customer Name"].head()
```

```python
df["Customer Name"] = (
    df["Customer Name"]
    .astype("string")
    .str.strip()
    .str.title()
)

df["Customer Name"].head()
```

## 6. Cleaning Company Names

I found unnecessary characters such as `-` and inconsistent
capitalization in company names.

I removed the unnecessary `-` and standardized the capitalization.

Bgauss is the correct spelling for the company name.

```python
df["Company"].unique()
```

```python
df["Company"] = (
    df["Company"]
    .astype("string")
    .str.strip()
    .str.replace("-", "", regex=False)
    .str.title()
)

# Correct company spelling
df["Company"] = df["Company"].replace(
    {
        "Bguss": "Bgauss",
        "Bgauss": "Bgauss"
    }
)

df["Company"].unique()
```

## 7. Cleaning Vehicle Base

I found different versions of the same vehicle base:

- ON ROAD
- ONROAD
- BASIC

I standardized them into:

- On Road
- Basic

```python
df["Vehical Base"].unique()
```

```python
df["Vehical Base"] = (
    df["Vehical Base"]
    .astype("string")
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.title()
)

df["Vehical Base"] = (
    df["Vehical Base"]
    .str.replace(
        r"^On\s*Road$",
        "On Road",
        case=False,
        regex=True
    )
    .str.replace(
        r"^Basic$",
        "Basic",
        case=False,
        regex=True
    )
)

df["Vehical Base"].unique()
```

## 8. Cleaning Vehicle Models

I standardized model names by removing unnecessary spaces
and applying consistent capitalization.

```python
df["Model"] = (
    df["Model"]
    .astype("string")
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.title()
)

df["Model"].unique()
```

## 9. Standardizing Payment Type

I found different capitalization such as CASH, Cash, LOAN and Loan.

I standardized them to uppercase so that the values are consistent.

```python
df["Payment Type"].unique()
```

```python
df["Payment Type"] = (
    df["Payment Type"]
    .astype("string")
    .str.strip()
    .str.upper()
)

df["Payment Type"].unique()
```

## 10. Cleaning Mobile Numbers

I removed unnecessary characters such as spaces, `+`, `-`
and brackets from mobile numbers.

```python
df["Mobile No."] = (
    df["Mobile No."]
    .astype("string")
    .str.replace(r"\D", "", regex=True)
)

df["Mobile No."].head()
```

## 11. Cleaning Vehicle Engine Numbers

The Vehicle Engine Number column contains a combination of
letters and numbers.

I only removed `-` characters and kept the letters and numbers
unchanged because these values are identifiers.

```python
df["Vehical Engine No."] = (
    df["Vehical Engine No."]
    .astype("string")
    .str.strip()
    .str.replace("-", "", regex=False)
)
```

## 12. Cleaning Vehicle Chassis Numbers

The chassis number is also an identifier, so I removed only
unnecessary `-` characters without changing the actual values.

```python
df["Vehical Chassis No."] = (
    df["Vehical Chassis No."]
    .astype("string")
    .str.strip()
    .str.replace("-", "", regex=False)
)
```

## 13. Working with Date Columns

I converted Booking Date and Date of Birth into Pandas datetime
format so they can be used properly for future analysis.

```python
df["Booking Date"] = pd.to_datetime(
    df["Booking Date"],
    errors="coerce"
)

df["Date of Birth"] = pd.to_datetime(
    df["Date of Birth"],
    errors="coerce"
)

df[["Booking Date", "Date of Birth"]].head()
```

## 16. Final Data Check

Before exporting the final dataset, I performed one last cleaning
and validation step.

The `Disbursed Date` and `Disbursed Amount` columns were removed
because they were not required for this project.

I also corrected the `Bgus` spelling in the Model column to `Bgauss`
while keeping the rest of each model name unchanged.

For example:
- Bgus A2 → Bgauss A2
- Bgus A8 → Bgauss A8
- Bgus B2 → Bgauss B2

Finally, I checked the cleaned dataset before exporting it.

```python
# Remove unnecessary columns
df = df.drop(
    columns=["Disbursed Date", "Disbursed Amount"],
    errors="ignore"
)

# Correct Bgus spelling in every Model value
# Keeps the rest of the model unchanged
df["Model"] = (
    df["Model"]
    .astype("string")
    .str.replace(
        r"Bgus",
        "Bgauss",
        case=False,
        regex=True
    )
)

# Final check
print("Dataset Shape:", df.shape)

print("\nModel values:")
print(df["Model"].unique())

print("\nRemaining empty cells:")
print(df.isnull().sum().sum())

df.head()
```

```python
df["Colour"] = (
    df["Colour"]
    .astype("string")
    .str.replace(
        r"Biue",
        "Blue",
        case=False,
        regex=True
    )
)
```

```python
# ============================================================
# FINAL CORRECTIONS
# ============================================================

# 1. Colour: biack → Black
df["Colour"] = df["Colour"].replace(
    r"Biack",
    "Black",
    regex=True
)

# 2. Model corrections
# Bajajct → Bajaj CT
# Apachi → Apache
df["Model"] = (
    df["Model"]
    .astype("string")
    .str.replace(
        r"Bajajct",
        "Bajaj CT",
        case=False,
        regex=True
    )
    .str.replace(
        r"Apachi",
        "Apache",
        case=False,
        regex=True
    )
)

# 3. Vehicle Engine Number:
# Remove "-" but keep letters and numbers unchanged
df["Vehical Engine No."] = (
    df["Vehical Engine No."]
    .astype("string")
    .str.replace("-", "", regex=False)
)

# 4. Date of Birth:
# Convert empty/missing values to "Not Provided"
df["Date of Birth"] = df["Date of Birth"].replace(
    ["", " ", "NaN", "nan", "NaT"],
    "Not Provided"
)

df["Date of Birth"] = df["Date of Birth"].fillna("Not Provided")


# ============================================================
# CHECK THE FINAL CHANGES
# ============================================================

print("Colour values:")
print(df["Colour"].unique())

print("\nModel values:")
print(df["Model"].unique())

print("\nDate of Birth:")
print(df["Date of Birth"].unique())

print("\nVehicle Engine Numbers:")
print(df["Vehical Engine No."].head())
```
